# DQL from Python

Uses the Python `dql` library (`Engine`) against **DynamoDB Local**.

Start Lab with Local settings so magics pick up the same endpoint:

```bash
./notebook/start.sh --local --start-local
```

Do not commit executed output from a live AWS account.

````python
from dql import Engine

engine = Engine()
engine.connect(
    "us-west-1",
    host="localhost",
    port=8000,
    is_secure=False,
    access_key="fakeid",
    secret_key="fakekey",
)
print(engine.region)
````

````python
engine.execute(
    """
    DROP TABLE IF EXISTS nb_posts;
    CREATE TABLE nb_posts (
      username STRING HASH KEY,
      postid NUMBER RANGE KEY,
      ts NUMBER INDEX('ts-index'),
      THROUGHPUT (5, 5)
    );
    INSERT INTO nb_posts (username, postid, ts, text) VALUES
      ('steve', 1, 1386413481, 'Hey guys!'),
      ('steve', 2, 1386413516, 'Guys?'),
      ('drdice', 1, 1386413575, 'No one here');
    """
)
````

````python
engine.execute(
    "SELECT * FROM nb_posts WHERE username = 'steve'",
    pretty_format=True,
)
````
